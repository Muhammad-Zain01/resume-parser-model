# 01 — Resume data processing

Extract usable text from normalized resume PDFs and DOCX files, then identify image-based files for later OCR.

**Input:** `.data/raw/` (with a temporary fallback to `.data/pdf/`)

**Output:** `.data/extracted/`

```python
from pathlib import Path

PROJECT_ROOT = Path.cwd()
RAW_DIR = PROJECT_ROOT / ".data" / "raw"
EXTRACTED_DIR = PROJECT_ROOT / ".data" / "extracted"

print(f"Raw directory: {RAW_DIR}")
print(f"Extracted directory: {EXTRACTED_DIR}")
```

## Processing plan

1. Discover every PDF and DOCX under the raw directory.
2. Extract PDF text with `pypdf` and DOCX text with `python-docx`.
3. Apply safe Unicode and whitespace cleanup.
4. Write `[IMAGE_BASED]` for documents that contain images but no text.
5. Mirror the category folders under the extracted directory.
6. List marker files for a later OCR pass.
